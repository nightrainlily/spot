```python
from bs4 import BeautifulSoup
import pandas as pd
import plotly.graph_objects as go
```

```python
with open('resources/enao.html', 'r', encoding='utf-8') as file:
    html_content = file.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

```python
genres = soup.find(class_='canvas')
print(genres.text)
```

```python
genre_map = {}

for genre in genres.find_all(recursive=True):
    name = genre.text[:-2]
    style = genre.get('style', '')
    styles = dict(item.split(':') for item in style.split(';') if item.strip())
    top = styles.get(' top', '').strip()
    left = styles.get(' left', '').strip()
    color = styles.get('color', '').strip()
    genre_map[name] = [top, left, color]

genre_map = {k: v for k, v in genre_map.items() if not k == ''}
```

```python
genre_map
```

```python
data = [
    {'genre': genre, 'x': int(values[0][:-2]), 'y': int(values[1][:-2]), 'color': values[2]}
    for genre, values in genre_map.items()
]

df = pd.DataFrame(data)
df.head()
```

```python
fig = go.Figure()

for item in data:
    fig.add_trace(
        go.Scatter(
            x=[item['x']],
            y=[item['y']],
            mode='markers+text',
            marker=dict(color=item['color'], size=5),
            # text=item['genre'],
            # textposition="top center",
            name=item['genre'],
            opacity=0.5
        )
    )

fig.update_layout(
    title="Genre Scatter Plot",
    xaxis_title='← denser & atmospheric, spikier & bouncier →',
    yaxis_title='← organic, mechanical & electric →',
    showlegend=True
)

fig.show()
```

```python
def get_coords(genre):
    return df[df['genre'] == genre]['x'].values[0], df[df['genre'] == genre]['y'].values[0]

```

```python
playlist_genres = ['indie pop', 'modern indie folk', 'hyperpop', 'hypnagogic pop']

shape = []
for genre in playlist_genres:
    shape.append(get_coords(genre))

shape = list(zip(*shape))
shape
```

```python
fig.add_trace(
    go.Scatter(
        x=shape[0],
        y=shape[1],
        fill='toself'
    )
)

fig.show()
```
